```python
import numpy as np
```

```python
X=np.array([
    [1,0,1,0],
    [1,0,1,1],
    [0,1,0,1]
])
y=np.array([[1],[1],[0]])
```

```python
def sigmoid_activation(x):
    return 1/(1+np.exp(-x))

def derivative_sigmoid(x):
    return x*(1-x)
```

```python
input_neurons=X.shape[1]
hidden_neurons= 3
output_neurons=1
epochs=100000
learning_rate=0.1

#weights and bias initialization
wh=np.random.uniform(size=(input_neurons,hidden_neurons))
bh=np.random.uniform(size=(1,hidden_neurons))
wout=np.random.uniform(size=(hidden_neurons,output_neurons))
bout=np.random.uniform(size=(1,output_neurons))
```

```python
for i in range(epochs):
    #feedforward
    hidden_layer_input=np.dot(X,wh)+bh
    hidden_layer_activation=sigmoid_activation(hidden_layer_input)
    output_layer=np.dot(hidden_layer_activation,wout)+bout
    output=sigmoid_activation(output_layer)

    #Backpropogation
    output_loss= y-output
    output_slope=derivative_sigmoid(output)
    output_delta=output_loss * output_slope

    hidden_loss=output_delta.dot(wout.T)
    hidden_slope=derivative_sigmoid(hidden_layer_activation)
    hidden_delta=hidden_slope*hidden_loss

    wout=wout+hidden_layer_activation.T.dot(output_delta)* learning_rate
    wh=wh+X.T.dot(hidden_delta) * learning_rate
    bout=bout+np.sum(output_delta,axis=0,keepdims=True)*learning_rate
    bh=bh+np.sum(hidden_delta,axis=0,keepdims=True)*learning_rate
```
