```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences = [
    "I really likeed the movie",
    "the film was amazing",
    "i like the first part but not second",
    "movie was terrible",
    "i really enjoyed the movie",
    "the movie was boring",
    "i hate that film"
]
labels = np.array([1,1,1,0,1,0,0]) # 1 for positive, 0 for negative
```

```python
#create object of Tokenizer
tokenizer=Tokenizer()

#tokenize the sentence into words 
tokenizer.fit_on_texts(sentences)
```

```python
#convert text ->numbers 
sequences = tokenizer.texts_to_sequences(sentences)
```

```python
print(sequences)
```

```python
#make all sequences same length by adding 0
x = pad_sequences(sequences)
```

```python
vocal_size = len(tokenizer.word_index) + 1
max_length = x.shape[1]
```

```python
model=Sequential([
    #Embedding layer 
    Embedding(input_dim=vocal_size,output_dim=10,input_length=max_length),
    #LSTM layer(sequence understanding with gates)
    LSTM(32),
    Dense(1,activation="sigmoid")

])
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
model.fit(x,labels,epochs=100)
```

```python
def predict_sentiment(text):
    seq=tokenizer.texts_to_sequences([text])
    padded=pad_sequences(seq,maxlen=max_length)
    prediction=model.predict(padded)
    print(prediction)
    return "Positive" if prediction[0][0]>0.5 else "Negative"
```

```python
test_sentence = "movie was amazing"
prediction=predict_sentiment(test_sentence)
print(prediction)
```
