```python
import cv2
```

```python
cap=cv2.VideoCapture(0)
while True:
    #get frames from video
    ret, frame=cap.read()
    #break if the video ends of fails
    if not ret:
        break
    cv2.imshow("My camera", frame)
    key=cv2.waitKey(25) & 0xFF
    if key == 27:
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import yt_dlp
```

```python
URL="https://youtu.be/YyepU5ztLf4?si=Dy9jRW0k7rm-bh8V"
ydl_options={
    'format': 'best[ext=mp4]'
}
with yt_dlp.YoutubeDL(ydl_options) as ydl:
    info=ydl.extract_info(URL, download=False)
    #Get video stream URL
    video_url=info['url']
    cap=cv2.VideoCapture(video_url)
    while True:
        #get frames from video
        ret, frame=cap.read()
        #break if the video ends of fails
        if not ret:
            break
        cv2.imshow("My camera", frame)
        key=cv2.waitKey(25) & 0xFF
        if key == 27:
            break
    cap.release()
    cv2.destroyAllWindows()
```
