```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('doc.png')
```

```python
img.shape
```

```python
rows,cols=img.shape[:2]
```

```python
plt.imshow(img)
plt.show()
```

```python
# Create translation matrix
#Format : [[1,0,tx],[0,1,ty]]
#tx- shift along x-axis(right)
#ty- shift along y-axis(down)
matrix=np.float32([[1,0,10],[0,1,50]])

#Apply translation
translated = cv2.warpAffine(img, matrix, (cols, rows))
plt.imshow(translated)
plt.show()
```

```python
#Rotation
#define center of rotation
center=(cols//2,rows//2)
```

```python
# Parameters: center, angle, scale
#angle- angle of rotation in degree(positive-anticlockwise, negative-clockwise)
#scale - zoom (1- original size, >1- zoom in, <1- zoom out)
matrix=cv2.getRotationMatrix2D(center, 45, 1)

#Apply rotation
rotated=cv2.warpAffine(img, matrix, (cols, rows))
plt.imshow(rotated)
plt.show()
```

```python
#Resize the image
#Parameters: image, (weidth, height)
resized=cv2.resize(img, (160, 160))
plt.imshow(resized)
plt.show()
```

```python
#Perspective Transformation
#Concept: tilted document  -> top down flat view

# *format= define 4 source points(4 corners of document in original image) 
#top left, top right, bottom right, bottom left
points1=np.float32([[32,15],[70,25],[85,61],[53,70]])

#Define destination points (4 corners of output image)
points2=np.float32([[0,0],[40,0],[40,60],[0,60]])

matrix=cv2.getPerspectiveTransform(points1, points2)

#Apply perspective transformation
result=cv2.warpPerspective(img, matrix, (40,60))
plt.imshow(result)
plt.show()
```
