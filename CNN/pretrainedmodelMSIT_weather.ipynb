```python
import tensorflow as tf
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Model
```

```python
# Load the pretrained weather classification model
preModel = tf.keras.models.load_model('../CASE_STUDY_WEATHER_RECOGNITION/weather_classification_model.h5')
```

```python
preModel.summary()
```

```python
# Optional: Freeze the layers of the pretrained model so their weights don't update during training
for layer in preModel.layers:
    layer.trainable = False

# Keras 3 mein Sequential model se direct .output lene par kabhi kabhi tensor error aata hai.
# Isliye hum Functional API ke bajaye seedha Sequential wrapper use kar sakte hain:
model = tf.keras.Sequential([
    preModel,
    Dense(1024, activation='relu'),
    Dense(10, activation='softmax')
])
```
