```python
text="I m learning python programming and python is fast "
print("Original text: ",text)
```

 #Text processing pipeline 
 #text ->Cleaning -> Tokenization->Stopwords Removal 

```python
#Step-1: Convert to lowercase
text=text.lower()
print("LowerCase: ",text)
```

```python
#Step-2: Remove Punctuation -!@#$%^&*()_-'',./
#re-regex-regular expression
import re
text=re.sub(r'[^\w\s]','',text)
#remove evrything expect letters and spaces 
print("Without punctuation:" ,text)
```

```python
#r file k path ko shi krta h means \u open nhi hota h kabhi toh usse \\n bna deta h 
```

```python
#step-3:Tokenization(Split sentence into words)
tokens= text.split()
print("Tekens: ",tokens)
```

```python
#Step-4: Remove Stopwords - is,am,are,the,a,an
stopwords=["a","m","is","i","the","and"]
filtered_tokens=[]
for word in tokens:
    if word not in stopwords:
        filtered_tokens.append(word)

print("After Removing stopwords:",filtered_tokens)
```

```python
#Step-5: stemming
def stem(word):
    if word.endswith("ing"):
        return word[:-3]
    return word
stemmed_words=[]
for word in filtered_tokens:
    stemmed_words.append(stem(word))

print("After stemming:",stemmed_words)
```

NLP Phase-2: WORD TO VECTOR

```python
#Will convert text -> count
from sklearn.feature_extraction.text import CountVectorizer
```

```python
sentences=["I am Learning python",
"We are happy today",
"I am sad today"]
```

```python
Vectorizer=CountVectorizer()
X=Vectorizer.fit_transform(sentences)
#learn vocublary
#convert sentences to numbers
```

```python
print("Vocublary: ",Vectorizer.get_feature_names_out())
#all unique words
#sorted in alphabetical order
#remove single letter like i,a
```

```python
print("Vectors: ")
print(X.toarray())
```

TF_IDF=Term Frequency- Inverse Documentary 

```python
#why tf-idf
#Beacuse count is not enough
#bag of words treat all words equally,but some words are more important than others
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
sentences=["I am Learning python",
"We are happy today",
"I am sad today"]
Vectorizer=TfidfVectorizer()
X=Vectorizer.fit_transform(sentences)
#build vacublary 
#calculate tf
#produces matrix
print("Vocublary: ",Vectorizer.get_feature_names_out())
print("TF_IDF Matrix: ")
print(X.toarray())
```

```python
#count +impact at each word 
#TF-IDF rewards rare important words and penalizes common words
```
