```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
text=["I am currently learning python programming","python programming is fun to learn","I am also learning java programming","I like java programming"]
```

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts(text)
```

```python
sequences = tokenizer.texts_to_sequences(text)
print(sequences)
```

```python
X=[]
y=[]
for seq in sequences:
    for i in range(1,len(seq)):
        X.append(seq[:i])
        y.append(seq[i])
```

```python
X
```

```python
X=pad_sequences(X)
```

```python
X
```

```python
y=np.array(y)
```

```python
vocab_size= len(tokenizer.word_index)+1
```

```python
model=Sequential([
    Embedding(vocab_size,20,input_length=X.shape[1]),
    SimpleRNN(320),
    Dense(vocab_size,activation='softmax')
])
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam')
```

```python
model.fit(X,y,epochs=100)
```

```python
input_text="i am learning"
seq=tokenizer.texts_to_sequences([input_text])[0]
seq=pad_sequences([seq],maxlen=X.shape[1]) 
# [] extra dimension laane k liye..upr data 2d m tha .. yha 1d m so .. isko bhi 2d m laana tha.
pred=model.predict(seq)
```

```python
pred
```

```python
pred_index=np.argmax(pred)
for word,index in tokenizer.word_index.items():
    if index == pred_index:
        print("Next Word: ",word)
```
