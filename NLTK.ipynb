```python
#Tokenization
from nltk.tokenize import word_tokenize
#Stopwords list
from nltk.corpus import stopwords
#Stemming
from nltk.stem import PorterStemmer
#lemmination
from nltk.stem import WordNetLemmatizer
```

```python
raw_text="I am learning python programming language. Python is greatest language..!!" 
```

```python
#Step 1: Lowercase
raw_text=raw_text.lower()
print("After Lowercasing: ",raw_text)

#Step-2: Tokenization
print(raw_text.split())
```

```python
tokens=word_tokenize(raw_text)
print("After Tokenization: ",tokens)
```

```python
#Step-3: Removing stopwords
english_stopwords = stopwords.words('english')
```

```python
filtered_tokens=[]
for word in tokens:
    if word not in english_stopwords:
        filtered_tokens.append(word)

print("After Removing Stopwords: ",filtered_tokens)
```

```python
#Step-4 Remove punctuation
import string
punctuations=string.punctuation
print("Punctuations: ",punctuations)
```

```python
clean_tokens=[word for word in filtered_tokens if word not in punctuations]
print("After Removing Punctuation: ",clean_tokens)
```

```python
stem=PorterStemmer()
#stem.stem("programming")
#stem.stem("programmer")
#stem.stem("program")
stem.stem("Cricket")
```

```python
wnet=WordNetLemmatizer()
wnet.lemmatize("programming","v")
```

```python
lemmatized_words=[]
for word in clean_tokens:
    lemmatized_words.append(wnet.lemmatize(word,"v"))
print("After Lemmatization: ",lemmatized_words)
```

```python
final_tokens=[]
for word in lemmatized_words:
    if word.isalpha():
        final_tokens.append(word)
print("After Removing Non-Alphabetic Tokens: ",final_tokens)
```

```python
cleaned_text=" ".join(final_tokens)
print("Cleaned Text: ",cleaned_text)
```
